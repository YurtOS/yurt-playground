# Suspend and resume a running notebook

This notebook's kernel is one CPython 3.14 process in a Yurt sandbox in this
browser tab, built so the sandbox can be **sealed**: the kernel's memory and
the Python process, stack included, written to this browser's IndexedDB.

Run the cell below. While it is printing primes, press **Suspend** in the
panel at the bottom right: the sandbox is sealed and torn down, and the
output stops. Press **Resume**: the sandbox comes back from the stored image
and the cell continues from exactly the prime it was on — same process, same
variables, same loop iteration.

```python
import time

primes = []
n = 1
while True:
    n += 1
    if all(n % p for p in primes if p * p <= n):
        primes.append(n)
        print(f"prime #{len(primes)} = {n}", flush=True)
        time.sleep(0.5)
```

Interrupt the cell (the ■ button) when you have seen enough; `primes` is
still there afterwards:
